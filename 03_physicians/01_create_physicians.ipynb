```python
import tabula
import pandas as pd
```

```python
# Read pdf into a list of DataFrame
dfs = tabula.read_pdf("data/Physicians_2026-05.pdf", pages='all')
```

```python
physicians_df = pd.concat(dfs)
physicians_df[physicians_df.MIDDLE_NAME.isnull()]
```

```python
physicians_df.SPECIALTY_DEF.value_counts(normalize=True)
```

```python
rehab_df = physicians_df[physicians_df.SPECIALTY_DEF == 'REHABILITATION MEDICINE']
rehab_df.PROVINCE.value_counts(normalize=True)
```
